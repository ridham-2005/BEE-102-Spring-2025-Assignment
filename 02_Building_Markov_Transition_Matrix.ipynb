# Q2) Building Markov transition matrix

The states here are the nucleotides: A, C, G, and T. The function computes the probability of transitioning from one nucleotide to the next.

Parameters: sequence (str): The input DNA sequence.

Returns: transition_matrix (dict): A dictionary representing the transition matrix. For example: transition_matrix['C']['G'] is the probability that G follows C.

```python
import pandas as pd

def markov_mat(seq: str) -> dict:
    states = ['A', 'C', 'G', 'T']
    cnts = {x: {y: 0 for y in states} for x in states}

    for i in range(len(seq) - 1):
        a, b = seq[i], seq[i + 1]
        if a in states and b in states:
            cnts[a][b] += 1

    mat = {x: {} for x in states}
    for x in states:
        total = sum(cnts[x].values())
        for y in states:
            mat[x][y] = cnts[x][y] / total if total > 0 else 0.0

    return mat
```

### Dinucleotide Transition Probability Calculation

1. **Define a List of Nucleotides**  
   The list contains the four DNA bases: `A`, `C`, `G`, `T`.

2. **Initialize Count Dictionary**  
   Create a nested dictionary `cnts` to store the count of each dinucleotide transition (e.g., how many times `A` is followed by `C`).

3. **Count Transitions**  
   Iterate through the DNA sequence one base at a time:
   - For each position `i`, increment the count of the transition from base `i` to base `i+1`.

4. **Convert Counts to Probabilities**  
   For each nucleotide `x`:
   - Compute the total number of transitions starting from `x`.
   - For each nucleotide `y`, compute the probability as:  
     `P(y | x) = count(x → y) / total(x)`
   - If `total(x) == 0`, assign a probability of `0.0` for all transitions from `x`.

5. **Result**  
   A dictionary of conditional probabilities representing the dinucleotide transition model.


```python
seq = "ACGTTCGTAGCATGATAAAATTTTTCCCCGGGGTGTCGAT"
res = markov_mat(seq)
```

Input sequence provided to build the Markov matrix.

```python
df = pd.DataFrame(res)
print("Markov Transition Matrix (rows: current state, columns: next state):")
df = df.T
print(df)
```
