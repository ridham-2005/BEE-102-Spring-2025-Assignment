# Q3) Multi-line sequence fasta to single-line fasta file


```python
import pandas as pd

def convert_fasta(input_file: str, output_file: str):
    with open(input_file, 'r') as inp, open(output_file, 'w') as out:
        found_header = False

        for i in inp:
            i = i.rstrip('\n')

            if i.startswith('>'):
                if found_header:
                    out.write('\n')
                out.write(i + '\n')
                found_header = True
            else:
                out.write(i.strip())

        out.write('\n')
```

### Function: FASTA File Formatter

This function processes a FASTA input file **line by line**.

---

### Header Lines (`>`)

- If a line **starts with `>`**, it's treated as a **new sequence header**.
- If it's **not the first header** (`found_header = True`), it writes a **newline** to separate from the previous sequence.
- Then, the header line is written **as-is**.

---

### Sequence Lines

- If the line is part of a sequence:
  - **Whitespace is stripped**
  - Characters are written **continuously**, without line breaks.

---

### Final Step

- Writes a **newline at the end** to ensure the output is **properly formatted**.

---
